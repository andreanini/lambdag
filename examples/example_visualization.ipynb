# Token-level LambdaG Visualization Example

This notebook demonstrates the `lambdaG_visualize` function from the `lambdag` package. 
LambdaG is a stylometric method for comparing writing styles at the token level.

## Parameters
- **q_text**: Questioned/disputed text (the text whose authorship is in question)
- **k_text**: Known text from a candidate author
- **ref_text**: Reference corpus text (background corpus)
- **N**: N-gram order for language models (default: 10)
- **r**: Number of replications for reference resampling (default: 30)
- **scale**: "absolute" (raw lambdaG) or "relative" (z-score)
- **negative**: If True, color negative lambdaG values in blue
- **rng**: Optional numpy random Generator for deterministic results

```python
# Example: Token-level LambdaG visualization
from lambdag.visualize import lambdaG_visualize
from IPython.display import HTML
import numpy as np

# Shakespeare text (questioned) - single sentence from King Lear
q_text = "Thus conscience doth make cowards of us all, and thus the native hue of resolution is sicklied o'er with the pale cast of thought."

# Shakespeare text (known) - 10 complete sentences from various plays
k_text = """To be or not to be that is the question.
Whether 'tis nobler in the mind to suffer the slings and arrows of outrageous fortune, or to take arms against a sea of troubles and by opposing end them.
All the world is a stage and all the men and women merely players.
Now is the winter of our discontent.
Brevity is the soul of wit.
The lady doth protest too much.
Something is rotten in the state of Denmark.
Parting is such sweet sorrow.
All that glitters is not gold.
The better part of valor is discretion.
What a piece of work is man."""

# Hemingway text (reference) - 11 complete sentences
ref_text = """The world breaks everyone and afterward many are strong at the broken places.
We ate well and simply and drank good red wine.
The sun also rises and the sun goes down.
He was an old man who fished alone in a skiff in the Gulf Stream.
The sea was very calm and the old man was happy.
He was an old man and he had gone eighty-four days without taking a fish.
The moon was low and the stars were bright.
He was a strange old man and he was very strong.
The fish moved with the current and the old man watched him.
It was a good morning and the sun was warm on his back.
He had seen many things and he had learned many things."""

# Compute visualization with deterministic seed
rng = np.random.default_rng(42)
result = lambdaG_visualize(
    q_text=q_text,
    k_text=k_text,
    ref_text=ref_text,
    N=5,  # smaller N for example
    r=10,  # fewer replications for speed
    output="html",
    scale="relative",
    negative=True,
    rng=rng,
)

# Display the HTML in an iframe to avoid CSS conflicts with notebook
HTML(
    f'<iframe srcdoc="{result["colourcoded_text"].replace(chr(34), "&quot;")}" '
    f'width="100%" height="600" style="border:none;"></iframe>'
)
```

## View the Results Table

The table contains per-token lambdaG scores and contributions.

```python
result["table"]
```

## Deterministic Output

Using the same seed produces identical results across runs:

```python
# Run twice with same seed - results should be identical
rng1 = np.random.default_rng(123)
rng2 = np.random.default_rng(123)

result1 = lambdaG_visualize(
    q_text=q_text,
    k_text=k_text,
    ref_text=ref_text,
    N=5,
    r=5,
    rng=rng1,
)

result2 = lambdaG_visualize(
    q_text=q_text,
    k_text=k_text,
    ref_text=ref_text,
    N=5,
    r=5,
    rng=rng2,
)

# Check if tables are identical
import pandas as pd

are_equal = result1["table"].equals(result2["table"])
print(f"Results are identical: {are_equal}")
```

## Example 2: Relative Scale (z-scores)

Using z-scores normalizes the lambdaG values, making it easier to compare across different texts.

```python
result_rel = lambdaG_visualize(
    q_text=q_text,
    k_text=k_text,
    ref_text=ref_text,
    N=5,
    r=10,
    output="html",
    scale="relative",  # use z-scores
    negative=True,
    rng=np.random.default_rng(42),
)

# Display the HTML in an iframe to avoid CSS conflicts with notebook
HTML(
    f'<iframe srcdoc="{result_rel["colourcoded_text"].replace(chr(34), "&quot;")}" '
    f'width="100%" height="600" style="border:none;"></iframe>'
)
```
